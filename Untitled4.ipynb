```python
import pandas as pd
import ipywidgets as widgets
from IPython.display import display, clear_output
from datetime import datetime

# -------------------------
# SHEET 1 - Main Dashboard
# -------------------------
# Task lists
personal_tasks = []
mustard_seed_tasks = []
art_to_hair_tasks = []
art_to_where_tasks = []

# Options
priority_options = ['Urgent', 'Mid Priority', 'Low Priority']
assigned_options = ['Lee', 'Cana', "Micha'el", 'Honey', 'Mike']
completed_options = ['No', 'Yes']

# Helper function to add tasks
def add_task(task_list, task, due_date, completed, priority=None, assigned=None):
    entry = {
        "Task": task,
        "Due Date": due_date,
        "Completed": completed
    }
    if priority:
        entry["Priority"] = priority
    if assigned:
        entry["Assigned To"] = ", ".join(assigned)
    task_list.append(entry)

# -------------------------
# Sheet 1 Widgets
# -------------------------
task_input = widgets.Text(description="Task:")
due_input = widgets.DatePicker(description="Due Date")
completed_input = widgets.Dropdown(options=completed_options, description="Completed")
notes_input = widgets.Textarea(description="Notes:", layout=widgets.Layout(width='50%'))
priority_input = widgets.Dropdown(options=priority_options, description="Priority")
assigned_input = widgets.SelectMultiple(options=assigned_options, description="Assigned To", layout=widgets.Layout(width='50%'))
category_dropdown = widgets.Dropdown(options=['Personal', 'Mustard Seed', 'Art to Hair', 'Art to Where'], description='Category:')
submit_btn = widgets.Button(description="Add Task", button_style="success")
delete_task_dropdown = widgets.Dropdown(options=[], description="Select Task to Delete")
delete_btn_sheet1 = widgets.Button(description="Delete Task", button_style='danger')
output_sheet1 = widgets.Output()

# -------------------------
# Sheet 1 Functions
# -------------------------
def update_sheet1_display():
    with output_sheet1:
        clear_output()
        # Update delete dropdown options
        delete_task_dropdown.options = [t["Task"] for t in personal_tasks + mustard_seed_tasks + art_to_hair_tasks + art_to_where_tasks]
        # Display each list
        if personal_tasks:
            display(widgets.HTML("<h3>Personal Tasks</h3>"))
            display(pd.DataFrame(personal_tasks).sort_values(by="Due Date"))
        if mustard_seed_tasks:
            display(widgets.HTML("<h3>Mustard Seed Tasks</h3>"))
            display(pd.DataFrame(mustard_seed_tasks).sort_values(by="Due Date"))
        if art_to_hair_tasks:
            display(widgets.HTML("<h3>Art to Hair Tasks</h3>"))
            display(pd.DataFrame(art_to_hair_tasks).sort_values(by="Due Date"))
        if art_to_where_tasks:
            display(widgets.HTML("<h3>Art to Where Tasks</h3>"))
            display(pd.DataFrame(art_to_where_tasks).sort_values(by="Due Date"))

def on_submit_clicked(b):
    category = category_dropdown.value
    task = task_input.value
    due = due_input.value
    completed = completed_input.value
    if category == "Personal":
        add_task(personal_tasks, task, due, completed)
    elif category == "Mustard Seed":
        add_task(mustard_seed_tasks, task, due, completed, priority_input.value, assigned_input.value)
    elif category == "Art to Hair":
        add_task(art_to_hair_tasks, task, due, completed)
    elif category == "Art to Where":
        add_task(art_to_where_tasks, task, due, completed)
    update_sheet1_display()
    update_sheet2_display()  # sync Sheet 2 in real-time

def delete_task_sheet1(b):
    selected = delete_task_dropdown.value
    global personal_tasks, mustard_seed_tasks, art_to_hair_tasks, art_to_where_tasks
    personal_tasks = [t for t in personal_tasks if t["Task"] != selected]
    mustard_seed_tasks = [t for t in mustard_seed_tasks if t["Task"] != selected]
    art_to_hair_tasks = [t for t in art_to_hair_tasks if t["Task"] != selected]
    art_to_where_tasks = [t for t in art_to_where_tasks if t["Task"] != selected]
    update_sheet1_display()
    update_sheet2_display()  # sync Sheet 2

submit_btn.on_click(on_submit_clicked)
delete_btn_sheet1.on_click(delete_task_sheet1)

# -------------------------
# SHEET 2 - To Do List
# -------------------------
to_do_list = []
in_progress = []
completed = []

output_sheet2 = widgets.Output()
task_dropdown_sheet2 = widgets.Dropdown(options=[], description="Select Task:")
move_in_progress_btn = widgets.Button(description="Move to In Progress", button_style='info')
move_completed_btn = widgets.Button(description="Move to Completed", button_style='success')
move_back_to_todo_btn = widgets.Button(description="Move Back to To Do", button_style='warning')
delete_completed_btn = widgets.Button(description="Delete from Completed", button_style='danger')

def sync_personal_tasks():
    global to_do_list
    current_tasks = [t["Task"] for t in in_progress + completed]
    to_do_list = sorted(
        [t for t in personal_tasks if t["Task"] not in current_tasks],
        key=lambda x: x["Due Date"] if x["Due Date"] else datetime.max
    )
    task_dropdown_sheet2.options = [t["Task"] for t in to_do_list + in_progress + completed]

def move_to_in_progress(b):
    selected = task_dropdown_sheet2.value
    task_item = next((t for t in to_do_list if t["Task"] == selected), None)
    if task_item:
        in_progress.append(task_item)
        to_do_list.remove(task_item)
    update_sheet2_display()

def move_to_completed(b):
    selected = task_dropdown_sheet2.value
    task_item = next((t for t in to_do_list if t["Task"] == selected), None)
    if task_item:
        completed.append(task_item)
        to_do_list.remove(task_item)
    update_sheet2_display()

def move_back_to_todo(b):
    selected = task_dropdown_sheet2.value
    task_item = next((t for t in in_progress if t["Task"] == selected), None)
    if task_item:
        to_do_list.append(task_item)
        in_progress.remove(task_item)
    update_sheet2_display()

def delete_from_completed(b):
    selected = task_dropdown_sheet2.value
    global completed, personal_tasks
    task_item = next((t for t in completed if t["Task"] == selected), None)
    if task_item:
        completed.remove(task_item)
        personal_tasks = [t for t in personal_tasks if t["Task"] != selected]
    update_sheet2_display()
    update_sheet1_display()

def update_sheet2_display():
    with output_sheet2:
        clear_output()
        sync_personal_tasks()
        display(widgets.HTML("<h3>To Do List</h3>"))
        display(pd.DataFrame(to_do_list).sort_values(by="Due Date") if to_do_list else "No tasks")
        display(widgets.HTML("<h3>In Progress</h3>"))
        display(pd.DataFrame(in_progress).sort_values(by="Due Date") if in_progress else "No tasks")
        display(widgets.HTML("<h3>Completed</h3>"))
        display(pd.DataFrame(completed).sort_values(by="Due Date") if completed else "No tasks")

# Connect Sheet 2 buttons
move_in_progress_btn.on_click(move_to_in_progress)
move_completed_btn.on_click(move_to_completed)
move_back_to_todo_btn.on_click(move_back_to_todo)
delete_completed_btn.on_click(delete_from_completed)

# -------------------------
# Display everything
# -------------------------
display(widgets.HTML("<h2>Sheet 1 - Main Dashboard</h2>"))
display(category_dropdown, task_input, due_input, completed_input, priority_input, assigned_input, notes_input, submit_btn)
display(delete_task_dropdown, delete_btn_sheet1)
display(output_sheet1)

display(widgets.HTML("<h2>Sheet 2 - To Do List (from Personal)</h2>"))
display(task_dropdown_sheet2, move_in_progress_btn, move_completed_btn, move_back_to_todo_btn, delete_completed_btn)
display(output_sheet2)

# Initial display
update_sheet1_display()
update_sheet2_display()
```
